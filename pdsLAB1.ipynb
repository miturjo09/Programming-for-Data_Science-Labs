```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

sum_result = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum =", sum_result)
print("Difference =", difference)
print("Product =", product)
print("Quotient =", quotient)


```

```python
c = float(input("Enter temperature in Celsius: "))

f = (c * 9/5) + 32
k = c + 273.15

print("Fahrenheit =", f)
print("Kelvin =", k)
```

```python
name = input("Enter student name: ")

m1 = float(input("Enter mark 1: "))
m2 = float(input("Enter mark 2: "))
m3 = float(input("Enter mark 3: "))
m4 = float(input("Enter mark 4: "))
m5 = float(input("Enter mark 5: "))

total = m1 + m2 + m3 + m4 + m5
average = total / 5

highest = max(m1, m2, m3, m4, m5)
lowest = min(m1, m2, m3, m4, m5)

passed = 0
if m1 >= 50:
    passed += 1
if m2 >= 50:
    passed += 1
if m3 >= 50:
    passed += 1
if m4 >= 50:
    passed += 1
if m5 >= 50:
    passed += 1

print("Student:", name)
print("Total =", total)
print("Average =", average)
print("Highest Mark =", highest)
print("Lowest Mark =", lowest)
print("Number of Passed Courses =", passed)
```
