```python
marks = [85, 67, 45, 90, 55]

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

passed = 0
for mark in marks:
    if mark >= 50:
        passed += 1

print("Total:", total)
print("Average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
print("Passed Courses:", passed)
```

```python
text = input("Enter a word or sentence: ")

characters = len(text)
vowels = 0
spaces = 0

for ch in text:
    if ch.lower() in "aeiou":
        vowels += 1
    if ch == " ":
        spaces += 1

print("Characters:", characters)
print("Vowels:", vowels)
print("Spaces:", spaces)
```

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

marks = [85, 75, 80, 90, 70]

grade = calculate_grade(marks)

print("Grade:", grade)
```

```python
products = [
    {"name": "Pen", "price": 10, "quantity": 5},
    {"name": "Book", "price": 120, "quantity": 2},
    {"name": "Bag", "price": 800, "quantity": 1},
    {"name": "Pencil", "price": 8, "quantity": 10}
]

total = 0

for product in products:
    total += product["price"] * product["quantity"]

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    return total

final_total = apply_discount(total)

print("Original Total:", total)
print("Final Total:", final_total)
```

```python
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total_marks = 0
passed_students = 0
highest_student = students[0]

print("Student Names and Marks:")
for student in students:
    print(student["name"], "-", student["marks"])

    total_marks += student["marks"]

    if student["marks"] >= 50:
        passed_students += 1

    if student["marks"] > highest_student["marks"]:
        highest_student = student

average = total_marks / len(students)

print("Passed Students:", passed_students)
print("Class Average:", average)
print("Highest Mark Holder:", highest_student["name"])
print("Highest Mark:", highest_student["marks"])
```

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total_inventory_value = 0
highest_product = products[0]
highest_line_total = highest_product["price"] * highest_product["quantity"]

print("Product Line Totals:")

for product in products:
    line_total = product["price"] * product["quantity"]

    print(product["name"], "=", line_total)

    total_inventory_value += line_total

    if line_total > highest_line_total:
        highest_line_total = line_total
        highest_product = product

print("Total Inventory Value:", total_inventory_value)
print("Highest Line Total Product:", highest_product["name"])
print("Highest Line Total:", highest_line_total)
```
