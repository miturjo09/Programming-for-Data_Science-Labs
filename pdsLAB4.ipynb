```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")

print("First 5 Rows:")
print(df.head())

print("Dataset Shape:")
print(df.shape)

print("Columns:")
print(df.columns)

print("Data Types:")
print(df.dtypes)

numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

print("nNumerical Attributes:")
print(list(numerical_cols))

print("nCategorical Attributes:")
print(list(categorical_cols))
```

```python
avg_fare = df.groupby("Pclass")["Fare"].mean()

plt.figure(figsize=(6,4))
plt.plot(avg_fare.index, avg_fare.values,
         marker='o', linewidth=2)
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["Age"], df["Fare"], alpha=0.5)
plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
gender_count = df["Sex"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(gender_count.index,
        gender_count.values,
        color=["red", "green"])

plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.hist(df["Age"].dropna(),
         bins=20,
         color="green",
         edgecolor="black")

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
survival_count = df["Survived"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    survival_count.values,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%",
    colors=["red", "green"]

)

plt.title("Survival Proportion")
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))

ax[0].hist(
    df["Age"].dropna(),
    bins=20,
    color="orange",
    edgecolor="black"
)
ax[0].set_title("Age Distribution")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")

class_count = df["Pclass"].value_counts().sort_index()

ax[1].bar(
    class_count.index.astype(str),
    class_count.values,
    color="steelblue"
)
ax[1].set_title("Passengers by Class")
ax[1].set_xlabel("Class")
ax[1].set_ylabel("Count")

plt.tight_layout()
plt.show()

```

```python
fig, ax = plt.subplots(1, 3, figsize=(18,5))

gender_count = df["Sex"].value_counts()
ax[0].bar(
    gender_count.index,
    gender_count.values,
    color=["blue", "orange"]
)
ax[0].set_title("Passengers by Gender")
ax[0].set_xlabel("Gender")
ax[0].set_ylabel("Count")

ax[1].hist(
    df["Fare"].dropna(),
    bins=20,
    color="pink",
    edgecolor="black"
)
ax[1].set_title("Fare Distribution")
ax[1].set_xlabel("Fare")
ax[1].set_ylabel("Frequency")

survival_count = df["Survived"].value_counts()
ax[2].pie(
    survival_count.values,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%",
    startangle=90
)
ax[2].set_title("Survival Proportion")

plt.tight_layout()
plt.show()
```
