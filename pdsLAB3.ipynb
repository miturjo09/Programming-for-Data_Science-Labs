```python
import pandas as pd


df = pd.read_csv("train.csv")
```

```python
display(df.head(10))

print("Shape (Rows, Columns):", df.shape)

print("Column Names:")
print(df.columns)

print("Data Types:")
print(df.dtypes)

display(df.describe())
```

```python
print("First Class Passengers:")
display(df[df["Pclass"] == 1].head())

print("Passengers younger than 18:")
display(df[df["Age"] < 18].head())

print("Male passengers who survived:")
display(df[(df["Sex"] == "male") & (df["Survived"] == 1)].head())

print("Passengers who paid a fare greater than 100:")
display(df[df["Fare"] > 100].head())
```

```python
print("10 passengers with the highest fares:")
display(df.sort_values(by="Fare", ascending=False).head(10)[["Name", "Fare"]])

print(f"Average passenger age: {df['Age'].mean():.2f}")

print("Passenger count by gender:")
print(df["Sex"].value_counts())

print("Average fare by passenger class:")
print(df.groupby("Pclass")["Fare"].mean())

print("Survival rate by passenger class:")
print(df.groupby("Pclass")["Survived"].mean())
```

```python
#How many passengers are in the dataset?
print(len(df))
#What is the average passenger age?
print(df["Age"].mean())
#Which passenger class has the highest number of passengers?
print(df["Pclass"].value_counts())
#What is the average fare paid by male and female passengers?
print(df.groupby("Sex")["Fare"].mean())
#Which gender has a higher survival rate?
print(df.groupby("Sex")["Survived"].mean())
#How does the survival rate differ across passenger classes?
print(df.groupby("Pclass")["Survived"].mean())
```

#Brief Findings

1.The dataset contains 891 passengers.

2.The average passenger age is 29.70 years.

3.Passenger Class 3 has the largest number of passengers.

4.Female passengers paid a higher average fare than male passengers.

5.Female passengers had a significantly higher survival rate than male passengers.

6.Survival rates were highest in First Class and lowest in Third Class, suggesting that ticket class influenced the likelihood of survival.
